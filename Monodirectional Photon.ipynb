```python
import openmc
```

```python
# Helper: convert MeV to eV
def mev_to_ev(energies_mev):
    return [e * 1e6 for e in energies_mev]
```

```python
# Define energy grid and PDF values (per MeV)
energies_per_mev   = [0.5, 1.0, 1.5]
pdf = [0.22, 0.5, 0.28]

#to convert MeV to eV
energies_ev = mev_to_ev(energies_per_mev)

#Create a Tabular energy distribution using linear (lin-lin) interpolation
energy_dist = openmc.stats.Tabular( energies_ev, pdf, interpolation='linear-linear' )

# Monodirectional angular distribution along +z (reference vector [0,0,1]) ,photons are emitted in asingle fixed direction
angle_dist = openmc.stats.Monodirectional([0.0, 0.0, 1.0])

#particle: "photon", "neutron", etc. 
#space: Spatial object
#angle: Angular object
#energy: Energy Distribution
mono_photon_src = openmc.Source()

#particle: determine the particle (neutron or photon)
mono_photon_src.particle = 'photon'

#Spatial distribution:Represents a point source   all particles originate from a single (x, y, z) location. point at the origin, point at the origin
mono_photon_src.space    = openmc.stats.Point()

mono_photon_src.angle    = angle_dist
mono_photon_src.energy   = energy_dist


openmc.settings.sources = [mono_photon_src]
```
