```python
import openmc
```

```python
# 1. Photon source, 1×10⁶ MeV, isotropic point
# Rationale: For a monoenergetic source we want a discrete function in energy; isotropic emission from a single point.
#particle: "photon", "neutron", etc. 
#space: Spatial object
#angle: Angular object
#energy: Energy Distribution
photon_source = openmc.IndependentSource()

#particle: determine the particle(photon or neutron)
photon_source.particle = 'photon'

#Spatial distribution:Represents a point source   all particles originate from a single (x, y, z) location. point at the origin
photon_source.space = openmc.stats.Point((0.0, 0.0, 0.0))

# Angular distribution: isotropic, Defines an isotropic angular distribution particles are emitted equally in all directions. over the unit sphere
photon_source.angle = openmc.stats.Isotropic()

# Energy distribution: Discrete, Defines a discrete energy distribution. Particles are emitted with specific energies according to assigned probabilities(1) with an energy of 1×10⁶ MeV & a probability of 1
photon_source.energy = openmc.stats.Discrete([1.0e12], [1.0])  # 1 × 10^6 MeV photons

openmc.settings.source = photon_source
```

```python
#2. Neutron source, 2.5×10⁻³ eV, isotropic point
# Rationale: identical approach for the neutron source energy and geometry (point).
neutron_src = openmc.IndependentSource()

#particle: determine the particle (neutron)
neutron_src.particle = 'neutron'

#Spatial distribution:Represents a point source   all particles originate from a single (x, y, z) location. point at the origin point at the origin
neutron_src.space    = openmc.stats.Point((0.0, 0.0, 0.0))

# Angular distribution: isotropic
neutron_src.angle    = openmc.stats.Isotropic()

# Energy distribution: Discrete with an energy of 2.5×10⁻³ eV & a probability of 1
neutron_src.energy   = openmc.stats.Discrete([2.5e-3], [1.0])

openmc.settings.source = neutron_src
```
